```python
# what is number data type 
# we have are three type of number
# 1. int
# 2. float
# 3. complex

print("\n --- Int data type ---")
a = 9
print("The value is: {} ".format(a))
print("The value is: {} , type of: {}".format(a, type(a)))
print("The value is: " + str(a) + " , type of: " + str(type(a)))
```

```python
print("\n --- Float data type ---")
b = 4.5
print("The value is {}, type of: {}".format(b, type(b)))
print("The value is: " + str(b) + " , type of: " + str(type(b)))
```

```python
print("\n ---- complex data type")
c= 5j
print("The value is {}, type of: {}". format(c, type(c)))
```

```python
x = 5
y =  2
res_add = x + y
res_sub = x - y
res_mul = x * y
res_div = x / y
res_modulus	 = x % y # باقیمانده
res_exponentiation = x ** y # توان
res_floor_division = x // y # خارج قسمت صحیح

print("x is: {} , y is:{}".format(x,y))
print("\nAddition result is: {}".format(res_add))
print("Subtraction result is: {}".format(res_sub))
print("Multiplication result is: {}".format(res_mul))
print("Division result is: {}".format(res_div))
print("Modulus result is: {}".format(res_modulus))
print("Exponentiation result is: {}".format(res_exponentiation))
print("Floor division result is: {}".format(res_floor_division))
```

```python
# --Precedence in mathematical operators
math_expression_01 = 9 - 2 * 3 # like is: res = (9 - (2 * 3))
math_expression_02 = 9 - 2 * 3 / 3 # like is: res = (9 - ((2 * 3) / 3))
math_expression_03 = (9 - 2) * 3 / 3 # like is: res =  (( (9 - 2) * 3) / 3 )

print("\n ---- Precedence in mathematical operators ----")
print("math_expression 01: {}".format(math_expression_01))
print("math_expression 02: {}".format(math_expression_02))
print("math_expression 03: {}".format(math_expression_03))
```
