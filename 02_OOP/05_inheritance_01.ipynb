```python
class Parent:      # کلاس والد (پایه)
    pass

class Child(Parent):  # کلاس فرزند (ارث‌بری از Parent)
    pass
```

```python
# کلاس والد
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id
    
    def display_info(self):
        print(f"نام: {self.name} | کد پرسنلی: {self.employee_id}")
    
    def calculate_salary(self):
        return 5000000  # حقوق پایه


# کلاس فرزند ۱: برنامه‌نویس
class Programmer(Employee):
    def __init__(self, name, employee_id, programming_language):
        super().__init__(name, employee_id)  # فراخوانی سازنده والد
        self.programming_language = programming_language
    
    def write_code(self):
        print(f"{self.name} در حال کدنویسی با {self.programming_language} است...")


# کلاس فرزند ۲: مدیر
class Manager(Employee):
    def __init__(self, name, employee_id, department):
        super().__init__(name, employee_id)
        self.department = department
    
    def manage_team(self):
        print(f"{self.name} تیم {self.department} را مدیریت می‌کند.")


# استفاده
prog = Programmer("علی", 101, "Python")
prog.display_info()
prog.write_code()

manager = Manager("مریم", 102, "فروش")
manager.display_info()
manager.manage_team()
```

```python
# Method Overriding
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    
    def start_engine(self):
        print(f"موتور {self.brand} {self.model} روشن شد. وروم وروم!")

class ElectricCar(Vehicle):
    # بازنویسی متد start_engine
    def start_engine(self):
        print(f"موتور برقی {self.brand} {self.model} فعال شد. (بی‌صدا!)")

class Motorcycle(Vehicle):
    # بازنویسی متد start_engine
    def start_engine(self):
        print(f"موتور {self.brand} {self.model} روشن شد. خر خر خر!")

# استفاده
car1 = Vehicle("تویوتا", "کمری")
car1.start_engine()

car2 = ElectricCar("تسلا", "Model 3")
car2.start_engine()

bike = Motorcycle("هوندا", "CBR")
bike.start_engine()
```

```python
# super()

class Animal:
    def __init__(self, name):
        self.name = name
    
    def eat(self):
        print(f"{self.name} در حال غذا خوردن است.")


class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)  # ابتدا سازنده والد را صدا می‌زنیم
        self.breed = breed      # سپس ویژگی جدید اضافه می‌کنیم
    
    def eat(self):
        super().eat()  # ابتدا رفتار والد را اجرا می‌کنیم
        print(f"{self.name} (نژاد {self.breed}) غذایش را تمام کرد.")


# استفاده
my_dog = Dog("رکس", "ژرمن شپرد")
my_dog.eat()
```
