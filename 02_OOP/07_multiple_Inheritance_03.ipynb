```python
class Flyer:
    def fly(self):
        print("در حال پرواز...")


class Swimmer:
    def swim(self):
        print("در حال شنا کردن...")


class Duck(Flyer, Swimmer):  # ارث‌بری از دو کلاس
    def quack(self):
        print("کواک کواک!")


# استفاده
duck = Duck()
duck.fly()    #  از Flyer ارث برده
duck.swim()   #  از Swimmer ارث برده
duck.quack()  #  متد مخصوص Duck
```

```python
class PaymentMethod:
    def __init__(self, amount):
        self.amount = amount
    
    def pay(self):
        raise NotImplementedError("زیرکلاس‌ها باید این متد را پیاده‌سازی کنند!")


class CreditCard(PaymentMethod):
    def __init__(self, amount, card_number):
        super().__init__(amount)
        self.card_number = card_number
    
    def pay(self):
        print(f"پرداخت {self.amount} تومان با کارت {self.card_number} انجام شد.")


class PayPal(PaymentMethod):
    def __init__(self, amount, email):
        super().__init__(amount)
        self.email = email
    
    def pay(self):
        print(f"پرداخت {self.amount} تومان از طریق PayPal ({self.email}) انجام شد.")


class Cash(PaymentMethod):
    def pay(self):
        print(f"پرداخت {self.amount} تومان به صورت نقدی انجام شد.")


# استفاده
payments = [
    CreditCard(500000, "6037-1234-5678-9012"),
    PayPal(300000, "user@example.com"),
    Cash(100000)
]

for payment in payments:
    payment.pay()
```

```python
# isinstance() و issubclass()
class Animal:
    pass

class Dog(Animal):
    pass

my_dog = Dog()

# بررسی نوع شیء
print(isinstance(my_dog, Dog))      # True
print(isinstance(my_dog, Animal))   # True (چون Dog از Animal ارث برده)
print(isinstance(my_dog, str))      # False

# بررسی رابطه وراثتی بین کلاس‌ها
print(issubclass(Dog, Animal))      # True
print(issubclass(Animal, Dog))      # False
```
