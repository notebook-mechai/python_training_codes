```python
# Defining the parent (base) class

class Animal:
    def __init__(self, name):
        self.name = name

    def eat(self):
        print(f"{self.name} در حال غذا خوردن است.")
        
class Dog(Animal):
    def bark(self):
        print(f"{self.name} می‌گوید: هاپ هاپ!")

class Cat(Animal):
    def meow(self):
        print(f"{self.name} می‌گوید: میو میو!")

class Bird(Animal):
    def fly(self):
        print(f"{self.name} در حال پرواز کردن است.")
        

# ساخت اشیاء و استفاده از کلاس‌ها
print("--- سگ ---")
my_dog = Dog("رکس")
my_dog.eat()
my_dog.bark()

print("\n--- گربه ---")
my_cat = Cat("پشمالو")
my_cat.eat()
my_cat.meow()

print("\n--- پرنده ---")
my_bird = Bird("طوطی")
my_bird.eat()
my_bird.fly()  
```
