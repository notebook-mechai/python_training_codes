```python
import numpy as np
```

```python
# 0-D Array

arr = np.array(4)
print(arr)
print(arr.ndim)
```

```python
# 1-D Array

arr_01 = np.array([11,22,33,44,55])
print(arr_01)
print(arr_01.ndim)
```

```python
# 2-D Array

arr_02 = np.array([[7,4,1], [8,5,2]])
print(arr_02)
print(arr_02.ndim)
```

```python
# 3-D Array
arr_03 = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
print(arr_03)
print(arr_03.ndim)
```

```python
# 5-D Array

arr_05 = np.array([1,2,3,4], ndmin=5)
print(arr_05)
print("number of dimension: ", arr_05.ndim)
```
