```python
def change_nums(a, b):
    temp = a
    a = b
    b = temp
    print("after swap numbers, a is: {} , b is: {}".format(a,b))

a = 12
b = 18
print("a is: {} , b is: {}".format(a,b))
change_nums(a, b)   
```

```python
def is_number(value):
    """بررسی می‌کند که آیا مقدار داده شده عدد است یا خیر"""
    # حالت اول: اگر خودش از نوع عدد باشد
    if isinstance(value, (int, float)):
        return True
    # حالت دوم: اگر رشته باشد، تلاش برای تبدیل
    if isinstance(value, str):
        try:
            float(value)
            return True
        except ValueError:
            return False
    # بقیه حالت‌ها
    return False


# تست‌ها
print(is_number(42))         # True   عدد صحیح
# print(is_number(3.14))       # True   عدد اعشاری
# print(is_number("100"))      # True   رشته عددی
# print(is_number("-7.5"))     # True   رشته عدد منفی اعشاری
# print(is_number("x"))        # False  حرف
# print(is_number("سلام"))     # False  کلمه
# print(is_number(None))       # False  مقدار خالی
```

```python
def get_numbers(flag = 0):
    if flag == 1:
        print("please enter again integer number\n")
    a = input("Enter a number:")
    b = input("Enter two number:")
    if (a.isdigit()) & (b.isdigit()):
        return a,b
    else:
        a = b = 0
        get_numbers(1)

def ch_nums(x,y):
    temp = x
    x = y
    y = temp
    return x,y


try:        
    (m,n) = get_numbers()
    # print("M: {} , N: {}".format(m,n) )
    (m,n) = ch_nums(x = m, y = n)
    print("m is: {} , n is: {}".format(m,n))
except Exception as e:
    print("Error:", e)
```
