```python
a1 = a = 2
b1 = b = 5
c1 = c = 4
d1 = d = 9
e1 = e = 12
f1 = f = 3
g1 = g = 2
# ***********
a += 1
b -= 1
c *= 1
d /= 1
e %= 1
f //= 1
g **= 2

print("a value is: {}, after 'a += 1'  is: {} ".format(a1,a))
print("a value is: {}, after 'b -= 1'  is: {} ".format(b1,b))
print("a value is: {}, after 'c *= 1'  is: {} ".format(c1,c))
print("a value is: {}, after 'd /= 1'  is: {} ".format(d1,d))
print("a value is: {}, after 'e %= 1'  is: {} ".format(e1,e))
print("a value is: {}, after 'f //= 1' is: {} ".format(f1,f))
print("a value is: {}, after 'g **= 2' is: {} ".format(g1,g))
```
