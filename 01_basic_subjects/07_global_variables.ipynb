```python
print("\n ---- Way one global variables ----")
word = "awesome"
def myFunc():
    print("python is " + word)

myFunc()
```

```python
print("\n ---- Way two global variables ---- ")
text_01 = "awesome"
def myFunc_02():
    text_01 = "fantastic"
    print("python is " + text_01)
 
myFunc_02()
```

```python
print("\n--------- Way two global variables")
text_02 = "awesome"
def myFunc_03():
    text_02 = "fantastic"
    print("python is " + text_02)
 
myFunc_03()
print("python is {}".format(text_02))
```

```python
print("\n--------- Way three global variables")
def myFunc_04():
    global msg
    msg = "fantastic"

myFunc_04()
print("python is " + msg)
```
