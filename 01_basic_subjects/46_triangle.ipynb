```python
rows = 6
for i in range(rows):
    for j in range(rows - i - 1):
        print(" ", end="")
    for k in range(2*i+1):
        print("*", end="")
    print()
```

```python
rs = 6
for x in range(rs):
    for y in range(rows - x - 1):
        print(" ", end="")
    for z in range(x+1):
        print("*", end="")
    print()
```
