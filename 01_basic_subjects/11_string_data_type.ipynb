```python
print("\n ---- concept f-string  ---- ")
name = "mor"
family = "Teza"
nick_name = "more"
print(f"{name}{family}")
print(f"My nick name is {nick_name}")
```

```python
text = "my name is {} {}"
print(text.format(name, family))
```

```python
text_2 = "my name is {a} {b}"
print(text_2.format(a = name, b = family))
```

```python
print("\n ---- concept index ----")

text_3 = "you are awesome"
#        "you are aw  e  s  o  m  e"
#      = "012 456 89 10 11 12 13 14"
print(text_3[9])
print(text_3[8:15])
print(text_3[-1])
print(text_3[8:20])
```

```python
print("\n ---- concept slicing index ----")

description = """
hello, how are you?
my name is john
"""
print(description)
print(description[1:]) # look like print(description)
print(description[:])
print(description[1:10])
print(description[-5:])
```

```python
print("\n ---- concept methods in string variable ---- ")

res_01 = description.count("a")
print(res_01)
```

```python
res_02 = description.find("o")
print("find o character:", res_02)
```

```python
res_03 = description.upper()
print("upper: ",res_03)
```

```python
res_04 = description.lower()
print("lower: ",res_04)
```

```python
res_05 = description.replace(",", "!")
print("replace , Instead of !: ",res_05)
```

```python
res_06 = description.title()
print("title: ",res_06)
```

```python
res_07 = description.isupper()
print("isupper:", res_07)
```

```python
res_08 = description.islower()
print("islower: ", res_08)
```

```python
res_09 = description.isnumeric()
print("isnumeric: ", res_09)
```

```python
res_10 = description.istitle()
print("istitle: ",res_10)
```

```python
res_11 = description.strip() # like trim in php
print("strip: ",res_11)
```

```python
res_12 = description.split(" ")
print("split: ", res_12)
```

```python
res_13 = description.capitalize() # just first chararcter to upper
print("capitalize: ",res_13)
```
