```python
# Nested Try-Except Blocks
def divide_and_access(a, b):
    try:
        result = a // b
        print("Result:", result)

        try:
            nums = [1, 2, 3]
            print(nums[5])

        except IndexError:
            print("Invalid index access")

    except ZeroDivisionError:
        print("Cannot divide by zero")

    finally:
        print("Execution completed")

divide_and_access(6, 2)
print("---")
divide_and_access(6, 0)
```
