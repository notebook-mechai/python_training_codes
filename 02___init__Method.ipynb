```python
# define class
class Person:
    def __init__(self, name, city,  age=65):
        self.name = name
        self.age = age
        self.city = city

# define object
name = input("you'r name:")
age = input("you'r age:")
city = input("you'r city:")
p = Person(name, city, age)

print("you'r name is {}, age is {}, city is {}".format(p.name, p.age, p.city))
```
