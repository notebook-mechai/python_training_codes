```python
print("\n ---- concept round ---- ")

pi = 3.141592653
res_01 = round(pi)
res_02 = round(pi, 2)
print("res 01:", res_01)
print("res 02:", res_02)
```

```python
num_1 = 20
num_2 = 18
num_3 = 14
avg = (num_1 + num_2 + num_3) / 3
res_03 = round(avg, 2)
print("simple avg: ", res_03)
print("addvance avg: ", res_03)
```

```python
print("\n ---- concept min ----")

res = min(num_1, num_2, num_3)
print(res)
```

```python
print("\n ---- concept MAX  ---- ")

res = max(num_1, num_2, num_3)
print(res)
```

```python
print("\n  ---- concept  absolute value ---- ")
num = -58
res = abs(num) # |-5| => 5 , |9| => 9
print(res)
```
