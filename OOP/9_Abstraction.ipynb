```python
from abc import ABC, abstractmethod

# کلاس انتزاعی (الگو)
class Shape(ABC):
    @abstractmethod
    def area(self):
        """این متد باید در زیرکلاس‌ها پیاده‌سازی شود"""
        pass
    
    @abstractmethod
    def perimeter(self):
        """این متد هم باید پیاده‌سازی شود"""
        pass
    
    def describe(self):
        """این متد معمولی است و اجباری نیست"""
        print(f"من یک {self.__class__.__name__} هستم")


# کلاس مربع
class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side ** 2
    
    def perimeter(self):
        return 4 * self.side


# کلاس دایره
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * (self.radius ** 2)
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius


# ❌ تلاش برای ساخت شیء از کلاس انتزاعی
# shape = Shape()  # TypeError: Can't instantiate abstract class Shape

# ✅ ساخت اشیاء از کلاس‌های فرزند
square = Square(5)
circle = Circle(3)

print(f"مساحت مربع: {square.area()}")           # 25
print(f"محیط مربع: {square.perimeter()}")        # 20
square.describe()                                # من یک Square هستم

print(f"مساحت دایره: {circle.area():.2f}")       # 28.27
print(f"محیط دایره: {circle.perimeter():.2f}")   # 18.85
circle.describe()                                # من یک Circle هستم
```

```python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def process_payment(self, amount):
        """پردازش پرداخت"""
        pass
    
    @abstractmethod
    def refund(self, transaction_id):
        """بازپرداخت"""
        pass
    
    def get_status(self):
        """متد معمولی (اجباری نیست)"""
        return "Active"


class CreditCardPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان با کارت اعتباری انجام شد.")
        return "TXN123456"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت برای تراکنش {transaction_id} انجام شد.")


class PayPalPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان از طریق PayPal انجام شد.")
        return "PP789012"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت PayPal برای تراکنش {transaction_id} انجام شد.")


class CryptoPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان با ارز دیجیتال انجام شد.")
        return "CRYPTO345678"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت کریپتو برای تراکنش {transaction_id} انجام شد.")


# استفاده
def checkout(payment_method, amount):
    """تابع پرداخت که با هر درگاه پرداختی کار می‌کند"""
    print(f"\n--- شروع پرداخت ---")
    transaction_id = payment_method.process_payment(amount)
    print(f"کد تراکنش: {transaction_id}")
    print(f"وضعیت درگاه: {payment_method.get_status()}")


# تست با روش‌های مختلف
checkout(CreditCardPayment(), 500000)
checkout(PayPalPayment(), 300000)
checkout(CryptoPayment(), 1000000)
```
