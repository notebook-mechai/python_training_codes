```python
from abc import ABC, abstractmethod

# کلاس انتزاعی (الگو)
class Shape(ABC):
    @abstractmethod
    def area(self):
        """این متد باید در زیرکلاس‌ها پیاده‌سازی شود"""
        pass
    
    @abstractmethod
    def perimeter(self):
        """این متد هم باید پیاده‌سازی شود"""
        pass
    
    def describe(self):
        """این متد معمولی است و اجباری نیست"""
        print(f"من یک {self.__class__.__name__} هستم")


# کلاس مربع
class Square(Shape):
    def __init__(self, side):
        self.side = side
    
    def area(self):
        return self.side ** 2
    
    def perimeter(self):
        return 4 * self.side


# کلاس دایره
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * (self.radius ** 2)
    
    def perimeter(self):
        return 2 * 3.14159 * self.radius


# ❌ تلاش برای ساخت شیء از کلاس انتزاعی
# shape = Shape()  # TypeError: Can't instantiate abstract class Shape

# ✅ ساخت اشیاء از کلاس‌های فرزند
square = Square(5)
circle = Circle(3)

print(f"مساحت مربع: {square.area()}")           # 25
print(f"محیط مربع: {square.perimeter()}")        # 20
square.describe()                                # من یک Square هستم

print(f"مساحت دایره: {circle.area():.2f}")       # 28.27
print(f"محیط دایره: {circle.perimeter():.2f}")   # 18.85
circle.describe()                                # من یک Circle هستم
```

```python
from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def process_payment(self, amount):
        """پردازش پرداخت"""
        pass
    
    @abstractmethod
    def refund(self, transaction_id):
        """بازپرداخت"""
        pass
    
    def get_status(self):
        """متد معمولی (اجباری نیست)"""
        return "Active"


class CreditCardPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان با کارت اعتباری انجام شد.")
        return "TXN123456"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت برای تراکنش {transaction_id} انجام شد.")


class PayPalPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان از طریق PayPal انجام شد.")
        return "PP789012"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت PayPal برای تراکنش {transaction_id} انجام شد.")


class CryptoPayment(PaymentGateway):
    def process_payment(self, amount):
        print(f"✅ پرداخت {amount} تومان با ارز دیجیتال انجام شد.")
        return "CRYPTO345678"
    
    def refund(self, transaction_id):
        print(f"✅ بازپرداخت کریپتو برای تراکنش {transaction_id} انجام شد.")


# استفاده
def checkout(payment_method, amount):
    """تابع پرداخت که با هر درگاه پرداختی کار می‌کند"""
    print(f"\n--- شروع پرداخت ---")
    transaction_id = payment_method.process_payment(amount)
    print(f"کد تراکنش: {transaction_id}")
    print(f"وضعیت درگاه: {payment_method.get_status()}")


# تست با روش‌های مختلف
checkout(CreditCardPayment(), 500000)
checkout(PayPalPayment(), 300000)
checkout(CryptoPayment(), 1000000)
```

```python
from abc import ABC, abstractmethod

class Authenticator(ABC):
    @abstractmethod
    def authenticate(self, username, password):
        """احراز هویت کاربر"""
        pass
    
    @abstractmethod
    def logout(self):
        """خروج کاربر"""
        pass
    
    def log_activity(self, action):
        """ثبت فعالیت (متد مشترک)"""
        print(f"[LOG] {action}")


class DatabaseAuthenticator(Authenticator):
    def authenticate(self, username, password):
        # شبیه‌سازی بررسی در دیتابیس
        if username == "admin" and password == "1234":
            self.log_activity(f"ورود موفق: {username}")
            return True
        self.log_activity(f"ورود ناموفق: {username}")
        return False
    
    def logout(self):
        self.log_activity("خروج کاربر")


class OAuthAuthenticator(Authenticator):
    def authenticate(self, username, password):
        # شبیه‌سازی احراز هویت با OAuth (مثل Google)
        self.log_activity(f"احراز هویت OAuth برای: {username}")
        return True  # فرض می‌کنیم همیشه موفق است
    
    def logout(self):
        self.log_activity("خروج از OAuth")


class LDAPAuthenticator(Authenticator):
    def authenticate(self, username, password):
        # شبیه‌سازی احراز هویت با LDAP (سازمانی)
        self.log_activity(f"احراز هویت LDAP برای: {username}")
        return username.startswith("emp_")  # فقط کارمندان
    
    def logout(self):
        self.log_activity("خروج از LDAP")


# استفاده
def login_system(auth_method, username, password):
    """سیستم ورود که با هر روش احراز هویتی کار می‌کند"""
    print(f"\n=== تلاش برای ورود با {auth_method.__class__.__name__} ===")
    if auth_method.authenticate(username, password):
        print("✅ ورود موفقیت‌آمیز!")
        auth_method.logout()
    else:
        print("❌ ورود ناموفق!")


# تست روش‌های مختلف
login_system(DatabaseAuthenticator(), "admin", "1234")
login_system(OAuthAuthenticator(), "user@gmail.com", "any")
login_system(LDAPAuthenticator(), "emp_123", "pass")
login_system(LDAPAuthenticator(), "guest", "pass")
```

```python
from abc import ABC, abstractmethod

class Storage(ABC):
    @abstractmethod
    def save(self, data):
        """ذخیره داده"""
        pass
    
    @abstractmethod
    def load(self):
        """بارگذاری داده"""
        pass
    
    @abstractmethod
    def delete(self, key):
        """حذف داده"""
        pass


class FileStorage(Storage):
    def __init__(self, filename):
        self.filename = filename
    
    def save(self, data):
        print(f"✅ ذخیره در فایل {self.filename}")
    
    def load(self):
        print(f"📖 بارگذاری از فایل {self.filename}")
        return {"data": "from file"}
    
    def delete(self, key):
        print(f"🗑️ حذف {key} از فایل")


class DatabaseStorage(Storage):
    def __init__(self, db_name):
        self.db_name = db_name
    
    def save(self, data):
        print(f"✅ ذخیره در دیتابیس {self.db_name}")
    
    def load(self):
        print(f"📖 بارگذاری از دیتابیس {self.db_name}")
        return {"data": "from database"}
    
    def delete(self, key):
        print(f"🗑️ حذف {key} از دیتابیس")


class CloudStorage(Storage):
    def __init__(self, bucket_name):
        self.bucket_name = bucket_name
    
    def save(self, data):
        print(f"☁️ ذخیره در فضای ابری {self.bucket_name}")
    
    def load(self):
        print(f"☁️ بارگذاری از فضای ابری {self.bucket_name}")
        return {"data": "from cloud"}
    
    def delete(self, key):
        print(f"☁️ حذف {key} از فضای ابری")


# استفاده
def data_manager(storage, data):
    """مدیریت داده که با هر سیستم ذخیره‌سازی کار می‌کند"""
    storage.save(data)
    loaded = storage.load()
    print(f"داده بارگذاری شده: {loaded}")


# تست سیستم‌های مختلف
data_manager(FileStorage("data.txt"), {"name": "علی"})
data_manager(DatabaseStorage("mydb"), {"name": "مریم"})
data_manager(CloudStorage("mybucket"), {"name": "رضا"})
```

```python
from abc import ABC, abstractmethod

class ReportGenerator(ABC):
    @abstractmethod
    def generate(self, data):
        """تولید گزارش"""
        pass
    
    @abstractmethod
    def export(self, filename):
        """خروجی گرفتن"""
        pass
    
    def format_data(self, data):
        """فرمت‌بندی داده (متد مشترک)"""
        return {k: str(v).upper() for k, v in data.items()}


class PDFReport(ReportGenerator):
    def generate(self, data):
        formatted = self.format_data(data)
        print(f"📄 تولید گزارش PDF با داده‌ها: {formatted}")
    
    def export(self, filename):
        print(f"💾 خروجی PDF در فایل {filename}.pdf")


class ExcelReport(ReportGenerator):
    def generate(self, data):
        formatted = self.format_data(data)
        print(f"📊 تولید گزارش Excel با داده‌ها: {formatted}")
    
    def export(self, filename):
        print(f"💾 خروجی Excel در فایل {filename}.xlsx")


class HTMLReport(ReportGenerator):
    def generate(self, data):
        formatted = self.format_data(data)
        print(f"🌐 تولید گزارش HTML با داده‌ها: {formatted}")
    
    def export(self, filename):
        print(f"💾 خروجی HTML در فایل {filename}.html")


# استفاده
def create_report(report_type, data, filename):
    """تولید گزارش با هر فرمتی"""
    report_type.generate(data)
    report_type.export(filename)


# تست فرمت‌های مختلف
sample_data = {"name": "علی", "age": 30, "city": "تهران"}

create_report(PDFReport(), sample_data, "report")
create_report(ExcelReport(), sample_data, "report")
create_report(HTMLReport(), sample_data, "report")
```
