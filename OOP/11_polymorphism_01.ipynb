```python
# مثال پیشرفته: سیستم پردازش فایل
"""
مزیت: اگر فردا بخواهیم PDFFile یا ExcelFile اضافه کنیم، نیازی به تغییر FileProcessor نیست!
"""
class TextFile:
    def __init__(self, filename):
        self.filename = filename
        self.content = "متن فایل تکست"
    
    def read(self):
        """خواندن فایل متنی"""
        return f"📄 محتوای {self.filename}: {self.content}"
    
    def process(self):
        """پردازش خاص فایل متنی"""
        return self.content.upper()


class ImageFile:
    def __init__(self, filename):
        self.filename = filename
        self.dimensions = (1920, 1080)
    
    def read(self):
        """خواندن فایل تصویری"""
        return f"🖼️ تصویر {self.filename} با ابعاد {self.dimensions}"
    
    def process(self):
        """پردازش خاص تصویر"""
        return f"تغییر اندازه به 50%"


class AudioFile:
    def __init__(self, filename):
        self.filename = filename
        self.duration = 180  # ثانیه
    
    def read(self):
        """خواندن فایل صوتی"""
        return f"🎵 فایل صوتی {self.filename} با مدت {self.duration} ثانیه"
    
    def process(self):
        """پردازش خاص صوت"""
        return "نرمال‌سازی صدا"


class VideoFile:
    def __init__(self, filename):
        self.filename = filename
        self.resolution = "4K"
    
    def read(self):
        """خواندن فایل ویدیویی"""
        return f"🎬 ویدیو {self.filename} با کیفیت {self.resolution}"
    
    def process(self):
        """پردازش خاص ویدیو"""
        return "فشرده‌سازی ویدیو"


# سیستم پردازش فایل (بدون دانستن نوع فایل!)
class FileProcessor:
    def __init__(self):
        self.files = []
    
    def add_file(self, file):
        """افزودن فایل (هر نوعی که باشد)"""
        self.files.append(file)
    
    def process_all(self):
        """پردازش تمام فایل‌ها"""
        print("="*60)
        print("🚀 شروع پردازش فایل‌ها")
        print("="*60)
        
        for file in self.files:
            # Duck Typing: فقط می‌دانیم متد read() و process() دارد
            print(file.read())
            result = file.process()
            print(f"   نتیجه پردازش: {result}")
            print("-"*60)


# استفاده
processor = FileProcessor()

processor.add_file(TextFile("document.txt"))
processor.add_file(ImageFile("photo.jpg"))
processor.add_file(AudioFile("music.mp3"))
processor.add_file(VideoFile("movie.mp4"))

processor.process_all()

```

```python
# نوع ۲: Method Overriding (بازنویسی متد)
# مثال ساده: سیستم حیوانات

class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        """متد پیش‌فرض"""
        return f"{self.name} صدایی ندارد."
    
    def move(self):
        """متد پیش‌فرض"""
        return f"{self.name} حرکت می‌کند."


class Dog(Animal):
    def speak(self):  # بازنویسی متد والد
        return f"{self.name} می‌گوید: هاپ هاپ!"
    
    def move(self):   # بازنویسی متد والد
        return f"{self.name} می‌دود."


class Cat(Animal):
    def speak(self):  # بازنویسی متد والد
        return f"{self.name} می‌گوید: میو میو!"
    
    def move(self):   # بازنویسی متد والد
        return f"{self.name} یواشکی راه می‌رود."


class Bird(Animal):
    def speak(self):
        return f"{self.name} می‌گوید: جیک جیک!"
    
    def move(self):
        return f"{self.name} پرواز می‌کند."


# استفاده از چندریختی
animals = [
    Dog("رکس"),
    Cat("پشمالو"),
    Bird("طوطی")
]

print("=== صداها ===")
for animal in animals:
    print(animal.speak())  # هر حیوان به روش خودش!

print("\n=== حرکت‌ها ===")
for animal in animals:
    print(animal.move())   # هر حیوان به روش خودش!
```

```python
# مثال پیشرفته: سیستم پرداخت
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    """کلاس پایه انتزاعی"""
    
    def __init__(self, amount):
        self.amount = amount
    
    @abstractmethod
    def process_payment(self):
        """پردازش پرداخت - باید در زیرکلاس‌ها پیاده‌سازی شود"""
        pass
    
    @abstractmethod
    def get_transaction_id(self):
        """دریافت شناسه تراکنش"""
        pass
    
    def get_receipt(self):
        """متد مشترک (بازنویسی نشده)"""
        return f"رسید پرداخت {self.amount} تومان"


class CreditCardPayment(PaymentMethod):
    def __init__(self, amount, card_number, cvv):
        super().__init__(amount)
        self.card_number = card_number
        self.cvv = cvv
    
    def process_payment(self):  # بازنویسی
        print(f"💳 پردازش پرداخت با کارت {self.card_number[-4:]}")
        # شبیه‌سازی پردازش
        import time
        time.sleep(0.5)
        return True
    
    def get_transaction_id(self):  # بازنویسی
        return f"CC-{hash(self.card_number) % 1000000}"


class PayPalPayment(PaymentMethod):
    def __init__(self, amount, email):
        super().__init__(amount)
        self.email = email
    
    def process_payment(self):  # بازنویسی
        print(f"🅿️ پردازش پرداخت از طریق PayPal ({self.email})")
        import time
        time.sleep(0.5)
        return True
    
    def get_transaction_id(self):  # بازنویسی
        return f"PP-{hash(self.email) % 1000000}"


class CryptoPayment(PaymentMethod):
    def __init__(self, amount, wallet_address):
        super().__init__(amount)
        self.wallet_address = wallet_address
    
    def process_payment(self):  # بازنویسی
        print(f"₿ پردازش پرداخت با کیف پول {self.wallet_address[:10]}...")
        import time
        time.sleep(1)  # کریپتو کندتر است
        return True
    
    def get_transaction_id(self):  # بازنویسی
        return f"CRYPTO-{hash(self.wallet_address) % 1000000}"


class BankTransferPayment(PaymentMethod):
    def __init__(self, amount, account_number):
        super().__init__(amount)
        self.account_number = account_number
    
    def process_payment(self):  # بازنویسی
        print(f"🏦 انتقال بانکی به حساب {self.account_number}")
        import time
        time.sleep(2)  # انتقال بانکی کندترین است
        return True
    
    def get_transaction_id(self):  # بازنویسی
        return f"BANK-{hash(self.account_number) % 1000000}"


# سیستم پرداخت (بدون دانستن نوع پرداخت!)
class PaymentProcessor:
    def __init__(self):
        self.transactions = []
    
    def checkout(self, payment_method):
        """پردازش پرداخت با هر روشی"""
        print("\n" + "="*60)
        print(f"💰 شروع پرداخت {payment_method.amount:,} تومان")
        print("="*60)
        
        # چندریختی: هر روش پرداخت به روش خودش پردازش می‌کند
        success = payment_method.process_payment()
        
        if success:
            transaction_id = payment_method.get_transaction_id()
            receipt = payment_method.get_receipt()
            
            self.transactions.append({
                'id': transaction_id,
                'amount': payment_method.amount,
                'method': payment_method.__class__.__name__
            })
            
            print(f"✅ پرداخت موفق!")
            print(f"📝 شناسه تراکنش: {transaction_id}")
            print(f"🧾 {receipt}")
        else:
            print("❌ پرداخت ناموفق!")
    
    def print_report(self):
        """گزارش تمام تراکنش‌ها"""
        print("\n" + "="*60)
        print("📊 گزارش تراکنش‌ها")
        print("="*60)
        for i, txn in enumerate(self.transactions, 1):
            print(f"{i}. {txn['method']}: {txn['amount']:,} تومان (ID: {txn['id']})")


# استفاده
processor = PaymentProcessor()

# پرداخت‌های مختلف
processor.checkout(CreditCardPayment(500000, "6037-1234-5678-9012", "123"))
processor.checkout(PayPalPayment(300000, "user@example.com"))
processor.checkout(CryptoPayment(1000000, "0x742d35Cc6634C0532925a3b844Bc9e7595f0bEb"))
processor.checkout(BankTransferPayment(750000, "IR8201234567890"))

# گزارش نهایی
processor.print_report()
```

```python
# نوع ۳: Operator Overloading (بارگذاری عملگرها)

"""
مفهوم
تغییر رفتار عملگرهای استاندارد پایتون (+, -, *, ==, <, > و...) برای کلاس‌های سفارشی.
 مثال ساده: کلاس مختصات
"""

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # بازنویسی عملگر + (جمع دو نقطه)
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    
    # بازنویسی عملگر - (تفریق دو نقطه)
    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)
    
    # بازنویسی عملگر * (ضرب در عدد)
    def __mul__(self, scalar):
        return Point(self.x * scalar, self.y * scalar)
    
    # بازنویسی عملگر == (مقایسه دو نقطه)
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    # بازنویسی عملگر < (مقایسه فاصله از مبدأ)
    def __lt__(self, other):
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)
    
    # بازنویسی print()
    def __str__(self):
        return f"({self.x}, {self.y})"


# استفاده
p1 = Point(3, 4)
p2 = Point(1, 2)

print(f"p1 = {p1}")              # (3, 4)
print(f"p2 = {p2}")              # (1, 2)
print(f"p1 + p2 = {p1 + p2}")    # (4, 6)   ← عملگر +
print(f"p1 - p2 = {p1 - p2}")    # (2, 2)   ← عملگر -
print(f"p1 * 3 = {p1 * 3}")      # (9, 12)  ← عملگر *
print(f"p1 == p2? {p1 == p2}")   # False    ← عملگر ==
print(f"p1 < p2? {p1 < p2}")     # False    ← عملگر <
```

```python
# مثال پیشرفته: کلاس ماتریس
class Matrix:
    def __init__(self, data):
        self.data = data
        self.rows = len(data)
        self.cols = len(data[0]) if data else 0
    
    # بازنویسی عملگر + (جمع دو ماتریس)
    def __add__(self, other):
        if self.rows != other.rows or self.cols != other.cols:
            raise ValueError("ابعاد ماتریس‌ها باید یکسان باشد!")
        
        result = []
        for i in range(self.rows):
            row = []
            for j in range(self.cols):
                row.append(self.data[i][j] + other.data[i][j])
            result.append(row)
        return Matrix(result)
    
    # بازنویسی عملگر * (ضرب دو ماتریس)
    def __mul__(self, other):
        if self.cols != other.rows:
            raise ValueError("تعداد ستون‌های اول باید برابر تعداد سطرهای دوم باشد!")
        
        result = [[0 for _ in range(other.cols)] for _ in range(self.rows)]
        
        for i in range(self.rows):
            for j in range(other.cols):
                for k in range(self.cols):
                    result[i][j] += self.data[i][k] * other.data[k][j]
        
        return Matrix(result)
    
    # بازنویسی عملگر * (ضرب در عدد)
    def __mul_scalar__(self, scalar):
        result = [[self.data[i][j] * scalar for j in range(self.cols)] 
                  for i in range(self.rows)]
        return Matrix(result)
    
    # بازنویسی عملگر == (مقایسه دو ماتریس)
    def __eq__(self, other):
        if self.rows != other.rows or self.cols != other.cols:
            return False
        for i in range(self.rows):
            for j in range(self.cols):
                if self.data[i][j] != other.data[i][j]:
                    return False
        return True
    
    # بازنویسی print()
    def __str__(self):
        result = ""
        for row in self.data:
            result += " | ".join(f"{x:3}" for x in row) + "\n"
        return result
    
    # بازنویسی repr()
    def __repr__(self):
        return f"Matrix({self.data})"


# استفاده
m1 = Matrix([[1, 2], [3, 4]])
m2 = Matrix([[5, 6], [7, 8]])

print("ماتریس ۱:")
print(m1)

print("ماتریس ۲:")
print(m2)

print("جمع دو ماتریس:")
print(m1 + m2)

print("ضرب دو ماتریس:")
print(m1 * m2)

print("ماتریس‌ها برابرند؟", m1 == m2)
```

```python
"""
📝 نوع ۴: Method Overloading (بارگذاری متد)
💭 مفهوم
تعریف چند متد با نام یکسان اما تعداد یا نوع آرگومان‌های متفاوت.
نکته مهم: پایتون به صورت مستقیم از Method Overloading پشتیبانی نمی‌کند! اما می‌توانیم با ترفندهایی آن را شبیه‌سازی کنیم.
🟢 مثال ساده: استفاده از مقادیر پیش‌فرض
"""

class Calculator:
    def add(self, a, b=0, c=0):
        """جمع 2 یا 3 عدد"""
        return a + b + c
    
    def multiply(self, *args):
        """ضرب هر تعداد عدد"""
        result = 1
        for num in args:
            result *= num
        return result


calc = Calculator()

print(calc.add(5, 3))        # 8   (2 آرگومان)
print(calc.add(5, 3, 2))     # 10  (3 آرگومان)
print(calc.add(5))           # 5   (1 آرگومان)

print(calc.multiply(2, 3))       # 6
print(calc.multiply(2, 3, 4))    # 24
print(calc.multiply(2, 3, 4, 5)) # 120
```

```python
# مثال پیشرفته: سیستم ارسال پیام

class MessageSender:
    def send(self, *args, **kwargs):
        """
        متد send که بسته به آرگومان‌ها، رفتار متفاوتی دارد
        """
        # حالت ۱: ارسال به یک نفر
        if len(args) == 2 and not kwargs:
            recipient, message = args
            return self._send_single(recipient, message)
        
        # حالت ۲: ارسال به چند نفر
        elif len(args) == 2 and 'recipients' in kwargs:
            message = args[0]
            recipients = kwargs['recipients']
            return self._send_multiple(recipients, message)
        
        # حالت ۳: ارسال با اولویت
        elif 'priority' in kwargs:
            recipient = args[0]
            message = args[1]
            priority = kwargs['priority']
            return self._send_with_priority(recipient, message, priority)
        
        # حالت ۴: ارسال زمان‌بندی شده
        elif 'schedule' in kwargs:
            recipient = args[0]
            message = args[1]
            schedule = kwargs['schedule']
            return self._send_scheduled(recipient, message, schedule)
        
        else:
            raise ValueError("آرگومان‌های نامعتبر!")
    
    def _send_single(self, recipient, message):
        """ارسال به یک نفر"""
        return f"📤 ارسال به {recipient}: {message}"
    
    def _send_multiple(self, recipients, message):
        """ارسال به چند نفر"""
        result = f"📤 ارسال گروهی به {len(recipients)} نفر:\n"
        for r in recipients:
            result += f"   - {r}\n"
        result += f"   پیام: {message}"
        return result
    
    def _send_with_priority(self, recipient, message, priority):
        """ارسال با اولویت"""
        emoji = "🔴" if priority == "high" else "🟡" if priority == "medium" else "🟢"
        return f"{emoji} ارسال با اولویت {priority} به {recipient}: {message}"
    
    def _send_scheduled(self, recipient, message, schedule):
        """ارسال زمان‌بندی شده"""
        return f"⏰ ارسال زمان‌بندی شده ({schedule}) به {recipient}: {message}"


# استفاده
sender = MessageSender()

# حالت ۱: ارسال ساده
print(sender.send("علی", "سلام!"))

# حالت ۲: ارسال گروهی
print(sender.send("پیام گروهی", recipients=["علی", "مریم", "رضا"]))

# حالت ۳: ارسال با اولویت
print(sender.send("مدیر", "گزارش فوری", priority="high"))

# حالت ۴: ارسال زمان‌بندی شده
print(sender.send("تیم", "جلسه فردا", schedule="2024-01-15 10:00"))
```
