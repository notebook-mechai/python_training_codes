```python
# کلاس‌های کاملاً مستقل (بدون وراثت!)
# مثال ۱: Duck Typing (ساده‌ترین نوع)
class Dog:
    def speak(self):
        return "هاپ هاپ! 🐕"

class Cat:
    def speak(self):
        return "میو میو! 🐱"

class Duck:
    def speak(self):
        return "کواک کواک! 🦆"

class Robot:
    def speak(self):
        return "بیپ بوپ! 🤖"

# تابعی که با هر شیئی کار می‌کند (به شرط داشتن متد speak)
def make_it_speak(animal):
    """این تابع نمی‌داند شیء از چه نوعی است!
    فقط می‌داند که متد speak() دارد."""
    print(animal.speak())


# تست
make_it_speak(Dog())    # هاپ هاپ! 🐕
make_it_speak(Cat())    # میو میو! 🐱
make_it_speak(Duck())   # کواک کواک! 🦆
make_it_speak(Robot())  # بیپ بوپ! 🤖  ← حتی ربات هم کار می‌کند!
```

```python
# مثال واقعی‌تر از Duck Typing
class List:
    def __init__(self, items):
        self.items = items
    
    def __len__(self):
        return len(self.items)

class Text:
    def __init__(self, content):
        self.content = content
    
    def __len__(self):
        return len(self.content)

# تابع len() پایتون با هر دو کار می‌کند!
my_list = List([1, 2, 3, 4, 5])
my_text = Text("سلام دنیا")

print(len(my_list))  # 5
print(len(my_text))  # 8
```

```python
# مثال ۲: Method Overriding (بازنویسی متد)
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        return f"{self.name} صدایی ندارد."
    
    def move(self):
        return f"{self.name} حرکت می‌کند."


class Dog(Animal):
    def speak(self):  # بازنویسی متد والد
        return f"{self.name} می‌گوید: هاپ هاپ!"
    
    def move(self):   # بازنویسی متد والد
        return f"{self.name} می‌دود."


class Cat(Animal):
    def speak(self):
        return f"{self.name} می‌گوید: میو میو!"
    
    def move(self):
        return f"{self.name} یواشکی راه می‌رود."


class Bird(Animal):
    def speak(self):
        return f"{self.name} می‌گوید: جیک جیک!"
    
    def move(self):
        return f"{self.name} پرواز می‌کند."


class Fish(Animal):
    def speak(self):
        return f"{self.name} نمی‌تواند صحبت کند! (حباب حباب)"
    
    def move(self):
        return f"{self.name} شنا می‌کند."


# جادوی چندریختی: یک حلقه، رفتارهای مختلف!
animals = [
    Dog("رکس"),
    Cat("پشمالو"),
    Bird("طوطی"),
    Fish("نمو")
]

print("=== صداها ===")
for animal in animals:
    print(animal.speak())  # هر حیوان به روش خودش!

print("\n=== حرکت‌ها ===")
for animal in animals:
    print(animal.move())   # هر حیوان به روش خودش!
```

```python
# مثال ۳: Operator Overloading (تغییر رفتار عملگرها)
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    # بازنویسی عملگر + (جمع دو بردار)
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)
    
    # بازنویسی عملگر - (تفریق دو بردار)
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)
    
    # بازنویسی عملگر * (ضرب در عدد)
    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)
    
    # بازنویسی عملگر == (مقایسه دو بردار)
    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
    
    # بازنویسی print()
    def __str__(self):
        return f"Vector({self.x}, {self.y})"


# استفاده
v1 = Vector(3, 4)
v2 = Vector(1, 2)

# حالا می‌توانیم بردارها را مثل اعداد جمع و تفریق کنیم!
print(v1 + v2)    # Vector(4, 6)   ← عملگر +
print(v1 - v2)    # Vector(2, 2)   ← عملگر -
print(v1 * 3)     # Vector(9, 12)  ← عملگر *
print(v1 == v2)   # False          ← عملگر ==
print(v1 == Vector(3, 4))  # True
```

```python
class Money:
    def __init__(self, amount, currency="تومان"):
        self.amount = amount
        self.currency = currency
    
    def __add__(self, other):
        if self.currency != other.currency:
            raise ValueError("نمی‌توان ارزهای مختلف را جمع کرد!")
        return Money(self.amount + other.amount, self.currency)
    
    def __sub__(self, other):
        if self.currency != other.currency:
            raise ValueError("نمی‌توان ارزهای مختلف را تفریق کرد!")
        result = self.amount - other.amount
        if result < 0:
            raise ValueError("موجودی کافی نیست!")
        return Money(result, self.currency)
    
    def __gt__(self, other):
        """عملگر >"""
        return self.amount > other.amount
    
    def __lt__(self, other):
        """عملگر <"""
        return self.amount < other.amount
    
    def __str__(self):
        return f"{self.amount:,} {self.currency}"


# استفاده
salary = Money(15000000)
rent = Money(5000000)
food = Money(3000000)

remaining = salary - rent - food
print(f"حقوق: {salary}")           # 15,000,000 تومان
print(f"اجاره: {rent}")             # 5,000,000 تومان
print(f"غذا: {food}")               # 3,000,000 تومان
print(f"باقی‌مانده: {remaining}")    # 7,000,000 تومان

print(f"حقوق > اجاره؟ {salary > rent}")  # True
```

```python
# مثال ۴: چندریختی با متدهای جادویی (Dunder Methods)
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages
    
    def __len__(self):
        """تعداد صفحات"""
        return self.pages
    
    def __str__(self):
        """نمایش زیبا"""
        return f"📖 {self.title} ({self.pages} صفحه)"
    
    def __repr__(self):
        """نمایش فنی"""
        return f"Book('{self.title}', {self.pages})"
    
    def __add__(self, other):
        """ترکیب دو کتاب"""
        return Book(f"{self.title} + {other.title}", self.pages + other.pages)
    
    def __eq__(self, other):
        """مقایسه بر اساس تعداد صفحات"""
        return self.pages == other.pages
    
    def __gt__(self, other):
        """کدام کتاب بلندتر است؟"""
        return self.pages > other.pages


# استفاده
book1 = Book("شاهنامه", 1200)
book2 = Book("دیوان حافظ", 500)
book3 = Book("مثنوی", 2500)

print(book1)              # 📖 شاهنامه (1200 صفحه)  ← __str__
print(len(book1))         # 1200                      ← __len__
print(book1 + book2)      # 📖 شاهنامه + دیوان حافظ (1700 صفحه)  ← __add__
print(book1 == book2)     # False                     ← __eq__
print(book3 > book1)      # True                      ← __gt__

# مرتب‌سازی کتاب‌ها (چون __gt__ را تعریف کردیم)
books = [book1, book2, book3]
sorted_books = sorted(books, key=lambda b: len(b))
for b in sorted_books:
    print(b)
```

```python
# مثال ۵: چندریختی در دنیای واقعی (سیستم اعلان‌ها)
class Notification(ABC):
    from abc import ABC, abstractmethod
    
    def __init__(self, recipient, message):
        self.recipient = recipient
        self.message = message
    
    @abstractmethod
    def send(self):
        """هر نوع اعلان به روش خودش ارسال می‌شود"""
        pass


class EmailNotification(Notification):
    def send(self):
        print(f"📧 ارسال ایمیل به {self.recipient}")
        print(f"   موضوع: {self.message}")
        print(f"   سرور: smtp.gmail.com")


class SMSNotification(Notification):
    def send(self):
        print(f"📱 ارسال پیامک به {self.recipient}")
        print(f"   متن: {self.message[:50]}...")
        print(f"   اپراتور: همراه اول")


class PushNotification(Notification):
    def send(self):
        print(f"🔔 ارسال نوتیفیکیشن به {self.recipient}")
        print(f"   پیام: {self.message}")
        print(f"   پلتفرم: Firebase")


class TelegramNotification(Notification):
    def send(self):
        print(f"✈️ ارسال پیام تلگرام به {self.recipient}")
        print(f"   متن: {self.message}")
        print(f"   بات: @MyBot")


# سیستم ارسال اعلان (بدون نیاز به دانستن نوع اعلان!)
class NotificationService:
    def __init__(self):
        self.notifications = []
    
    def add_notification(self, notification):
        self.notifications.append(notification)
    
    def send_all(self):
        print("="*50)
        print("🚀 شروع ارسال تمام اعلان‌ها")
        print("="*50)
        for notification in self.notifications:
            notification.send()  # ← جادوی چندریختی!
            print("-"*30)


# استفاده
service = NotificationService()

service.add_notification(EmailNotification("ali@gmail.com", "سفارش شما ثبت شد"))
service.add_notification(SMSNotification("09121234567", "کد تایید: 1234"))
service.add_notification(PushNotification("user_123", "پیام جدید دارید"))
service.add_notification(TelegramNotification("@ali_user", "به‌روزرسانی جدید"))

service.send_all()
```

```python
#مثال ۶: چندریختی با isinstance و type
class Shape:
    def area(self):
        return 0

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2

class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    
    def area(self):
        return 0.5 * self.base * self.height


# تابعی که مساحت کل را حساب می‌کند
def total_area(shapes):
    """بدون توجه به نوع شکل، مساحت کل را حساب می‌کند"""
    total = 0
    for shape in shapes:
        total += shape.area()  # ← چندریختی!
    return total


# استفاده
shapes = [
    Rectangle(5, 3),    # مساحت: 15
    Circle(4),          # مساحت: 50.27
    Triangle(6, 4),     # مساحت: 12
    Rectangle(10, 2),   # مساحت: 20
]

print(f"مساحت کل: {total_area(shapes):.2f}")  # 97.27
```
