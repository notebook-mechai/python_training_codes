```python
from abc import ABC, abstractmethod
from typing import List, Dict, Any

# کلاس انتزاعی برای پلاگین‌ها
class Plugin(ABC):
    @abstractmethod
    def initialize(self) -> bool:
        """راه‌اندازی اولیه پلاگین"""
        pass
    
    @abstractmethod
    def execute(self, data: Any) -> Any:
        """اجرای منطق اصلی پلاگین"""
        pass
    
    @abstractmethod
    def cleanup(self) -> None:
        """پاکسازی منابع"""
        pass
    
    @property
    @abstractmethod
    def name(self) -> str:
        """نام پلاگین"""
        pass
    
    @property
    @abstractmethod
    def version(self) -> str:
        """نسخه پلاگین"""
        pass


# پلاگین‌های مختلف
class DataValidatorPlugin(Plugin):
    @property
    def name(self) -> str:
        return "Data Validator"
    
    @property
    def version(self) -> str:
        return "1.0.0"
    
    def initialize(self) -> bool:
        print(f"✅ راه‌اندازی {self.name} v{self.version}")
        return True
    
    def execute(self, data: Dict) -> Dict:
        """اعتبارسنجی داده‌ها"""
        validated = {}
        for key, value in data.items():
            if isinstance(value, str) and len(value) > 0:
                validated[key] = value.strip()
            elif isinstance(value, (int, float)):
                validated[key] = value
            else:
                print(f"⚠️ داده نامعتبر برای کلید {key}")
        return validated
    
    def cleanup(self) -> None:
        print(f"🧹 پاکسازی {self.name}")


class DataTransformerPlugin(Plugin):
    @property
    def name(self) -> str:
        return "Data Transformer"
    
    @property
    def version(self) -> str:
        return "2.1.0"
    
    def initialize(self) -> bool:
        print(f"✅ راه‌اندازی {self.name} v{self.version}")
        return True
    
    def execute(self, data: Dict) -> Dict:
        """تبدیل داده‌ها"""
        return {k.upper(): v for k, v in data.items()}
    
    def cleanup(self) -> None:
        print(f"🧹 پاکسازی {self.name}")


class DataEncryptionPlugin(Plugin):
    @property
    def name(self) -> str:
        return "Data Encryption"
    
    @property
    def version(self) -> str:
        return "1.5.0"
    
    def initialize(self) -> bool:
        print(f"✅ راه‌اندازی {self.name} v{self.version}")
        return True
    
    def execute(self, data: Dict) -> Dict:
        """رمزنگاری داده‌ها (شبیه‌سازی)"""
        return {k: f"ENCRYPTED_{v}" for k, v in data.items()}
    
    def cleanup(self) -> None:
        print(f"🧹 پاکسازی {self.name}")


# سیستم مدیریت پلاگین
class PluginManager:
    def __init__(self):
        self.plugins: List[Plugin] = []
    
    def register_plugin(self, plugin: Plugin) -> None:
        """ثبت پلاگین"""
        if plugin.initialize():
            self.plugins.append(plugin)
            print(f"📦 پلاگین {plugin.name} ثبت شد")
    
    def process_data(self, data: Any) -> Any:
        """پردازش داده از طریق تمام پلاگین‌ها"""
        result = data
        for plugin in self.plugins:
            print(f"\n🔄 پردازش با {plugin.name}...")
            result = plugin.execute(result)
        return result
    
    def shutdown(self) -> None:
        """خاموش کردن تمام پلاگین‌ها"""
        for plugin in self.plugins:
            plugin.cleanup()


# استفاده
manager = PluginManager()

# ثبت پلاگین‌ها
manager.register_plugin(DataValidatorPlugin())
manager.register_plugin(DataTransformerPlugin())
manager.register_plugin(DataEncryptionPlugin())

# پردازش داده
sample_data = {"name": "علی", "age": 30, "city": "تهران"}
print("\n" + "="*50)
result = manager.process_data(sample_data)
print("="*50)
print(f"\nنتیجه نهایی: {result}")

# خاموش کردن
manager.shutdown()
```

```python
from abc import ABC, abstractmethod
from typing import List

# استراتژی انتزاعی
class SortStrategy(ABC):
    @abstractmethod
    def sort(self, data: List) -> List:
        """مرتب‌سازی داده‌ها"""
        pass


# استراتژی‌های مختلف
class BubbleSortStrategy(SortStrategy):
    def sort(self, data: List) -> List:
        print("🫧 مرتب‌سازی با Bubble Sort...")
        arr = data.copy()
        n = len(arr)
        for i in range(n):
            for j in range(0, n-i-1):
                if arr[j] > arr[j+1]:
                    arr[j], arr[j+1] = arr[j+1], arr[j]
        return arr


class QuickSortStrategy(SortStrategy):
    def sort(self, data: List) -> List:
        print("⚡ مرتب‌سازی با Quick Sort...")
        if len(data) <= 1:
            return data
        pivot = data[len(data) // 2]
        left = [x for x in data if x < pivot]
        middle = [x for x in data if x == pivot]
        right = [x for x in data if x > pivot]
        return self.sort(left) + middle + self.sort(right)


class MergeSortStrategy(SortStrategy):
    def sort(self, data: List) -> List:
        print("🔀 مرتب‌سازی با Merge Sort...")
        if len(data) <= 1:
            return data
        
        mid = len(data) // 2
        left = self.sort(data[:mid])
        right = self.sort(data[mid:])
        
        return self._merge(left, right)
    
    def _merge(self, left: List, right: List) -> List:
        result = []
        i = j = 0
        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1
        result.extend(left[i:])
        result.extend(right[j:])
        return result


# زمینه (Context) که از استراتژی استفاده می‌کند
class Sorter:
    def __init__(self, strategy: SortStrategy):
        self._strategy = strategy
    
    def set_strategy(self, strategy: SortStrategy) -> None:
        """تغییر استراتژی در زمان اجرا"""
        self._strategy = strategy
        print(f"🔄 استراتژی تغییر کرد به: {strategy.__class__.__name__}")
    
    def sort_data(self, data: List) -> List:
        """مرتب‌سازی با استراتژی فعلی"""
        return self._strategy.sort(data)


# استفاده
data = [64, 34, 25, 12, 22, 11, 90]

# ایجاد مرتب‌ساز با استراتژی Bubble Sort
sorter = Sorter(BubbleSortStrategy())
result1 = sorter.sort_data(data)
print(f"نتیجه: {result1}\n")

# تغییر استراتژی به Quick Sort
sorter.set_strategy(QuickSortStrategy())
result2 = sorter.sort_data(data)
print(f"نتیجه: {result2}\n")

# تغییر استراتژی به Merge Sort
sorter.set_strategy(MergeSortStrategy())
result3 = sorter.sort_data(data)
print(f"نتیجه: {result3}")
```

```python
from abc import ABC, abstractmethod
from typing import Dict, List, Any, Optional
import json

# مدل انتزاعی
class Model(ABC):
    @abstractmethod
    def save(self) -> None:
        """ذخیره در دیتابیس"""
        pass
    
    @abstractmethod
    def delete(self) -> None:
        """حذف از دیتابیس"""
        pass
    
    @classmethod
    @abstractmethod
    def find_by_id(cls, id: int) -> Optional['Model']:
        """یافتن بر اساس شناسه"""
        pass
    
    @classmethod
    @abstractmethod
    def all(cls) -> List['Model']:
        """دریافت تمام رکوردها"""
        pass


# دیتابیس انتزاعی
class Database(ABC):
    @abstractmethod
    def connect(self) -> None:
        """اتصال به دیتابیس"""
        pass
    
    @abstractmethod
    def disconnect(self) -> None:
        """قطع اتصال"""
        pass
    
    @abstractmethod
    def execute(self, query: str, params: tuple = ()) -> Any:
        """اجرای کوئری"""
        pass


# پیاده‌سازی دیتابیس فایل
class FileDatabase(Database):
    def __init__(self, filename: str):
        self.filename = filename
        self.data: Dict = {}
    
    def connect(self) -> None:
        try:
            with open(self.filename, 'r') as f:
                self.data = json.load(f)
        except FileNotFoundError:
            self.data = {}
        print(f"✅ اتصال به فایل {self.filename}")
    
    def disconnect(self) -> None:
        with open(self.filename, 'w') as f:
            json.dump(self.data, f, indent=2)
        print(f"✅ قطع اتصال از {self.filename}")
    
    def execute(self, query: str, params: tuple = ()) -> Any:
        # شبیه‌سازی عملیات دیتابیس
        if query == "INSERT":
            table, record = params
            if table not in self.data:
                self.data[table] = []
            self.data[table].append(record)
        elif query == "SELECT":
            table = params[0]
            return self.data.get(table, [])
        elif query == "DELETE":
            table, id = params
            if table in self.data:
                self.data[table] = [r for r in self.data[table] if r['id'] != id]


# مدل کاربر
class User(Model):
    def __init__(self, id: int, name: str, email: str, db: Database):
        self.id = id
        self.name = name
        self.email = email
        self.db = db
    
    def save(self) -> None:
        record = {'id': self.id, 'name': self.name, 'email': self.email}
        self.db.execute("INSERT", ('users', record))
        print(f"✅ کاربر {self.name} ذخیره شد")
    
    def delete(self) -> None:
        self.db.execute("DELETE", ('users', self.id))
        print(f"🗑️ کاربر {self.name} حذف شد")
    
    @classmethod
    def find_by_id(cls, id: int, db: Database) -> Optional['User']:
        users = db.execute("SELECT", ('users',))
        for user_data in users:
            if user_data['id'] == id:
                return cls(user_data['id'], user_data['name'], user_data['email'], db)
        return None
    
    @classmethod
    def all(cls, db: Database) -> List['User']:
        users = db.execute("SELECT", ('users',))
        return [cls(u['id'], u['name'], u['email'], db) for u in users]
    
    def __repr__(self):
        return f"User(id={self.id}, name='{self.name}', email='{self.email}')"


# استفاده
db = FileDatabase('users.json')
db.connect()

# ایجاد کاربران
user1 = User(1, "علی", "ali@example.com", db)
user2 = User(2, "مریم", "maryam@example.com", db)
user3 = User(3, "رضا", "reza@example.com", db)

# ذخیره
user1.save()
user2.save()
user3.save()

# دریافت همه کاربران
all_users = User.all(db)
print(f"\n📋 تمام کاربران: {all_users}")

# یافتن کاربر خاص
found_user = User.find_by_id(2, db)
print(f"\n🔍 کاربر یافت شده: {found_user}")

# حذف کاربر
user3.delete()

# قطع اتصال
db.disconnect()
```

```python
from abc import ABC, abstractmethod
from typing import Callable, Dict, Any
from functools import wraps

# Middleware انتزاعی
class Middleware(ABC):
    @abstractmethod
    def process_request(self, request: Dict) -> Dict:
        """پردازش درخواست"""
        pass
    
    @abstractmethod
    def process_response(self, response: Dict) -> Dict:
        """پردازش پاسخ"""
        pass


# Middleware‌های مختلف
class AuthenticationMiddleware(Middleware):
    def process_request(self, request: Dict) -> Dict:
        print("🔐 بررسی احراز هویت...")
        if 'token' not in request:
            raise PermissionError("❌ توکن احراز هویت موجود نیست!")
        request['user'] = "authenticated_user"
        return request
    
    def process_response(self, response: Dict) -> Dict:
        response['authenticated'] = True
        return response


class LoggingMiddleware(Middleware):
    def process_request(self, request: Dict) -> Dict:
        print(f"📝 ثبت درخواست: {request.get('path', '/')}")
        request['logged'] = True
        return request
    
    def process_response(self, response: Dict) -> Dict:
        print(f"📝 ثبت پاسخ: {response.get('status', 200)}")
        return response


class RateLimitMiddleware(Middleware):
    def __init__(self, max_requests: int = 100):
        self.request_count = 0
        self.max_requests = max_requests
    
    def process_request(self, request: Dict) -> Dict:
        self.request_count += 1
        print(f"⚡ بررسی محدودیت نرخ: {self.request_count}/{self.max_requests}")
        if self.request_count > self.max_requests:
            raise Exception("❌ محدودیت نرخ درخواست!")
        return request
    
    def process_response(self, response: Dict) -> Dict:
        response['rate_limit'] = f"{self.request_count}/{self.max_requests}"
        return response


class CacheMiddleware(Middleware):
    def __init__(self):
        self.cache: Dict = {}
    
    def process_request(self, request: Dict) -> Dict:
        path = request.get('path')
        if path in self.cache:
            print(f"💾 پاسخ از کش: {path}")
            request['cached'] = True
            request['cached_response'] = self.cache[path]
        else:
            print(f"💾 کش خالی برای: {path}")
        return request
    
    def process_response(self, response: Dict) -> Dict:
        path = response.get('path')
        if path and not response.get('cached'):
            self.cache[path] = response
            print(f"💾 ذخیره در کش: {path}")
        return response


# سرور با پشتیبانی از Middleware
class Server:
    def __init__(self):
        self.middlewares: list[Middleware] = []
    
    def add_middleware(self, middleware: Middleware) -> None:
        """افزودن Middleware"""
        self.middlewares.append(middleware)
        print(f"🔧 Middleware اضافه شد: {middleware.__class__.__name__}")
    
    def handle_request(self, request: Dict) -> Dict:
        """پردازش درخواست"""
        print("\n" + "="*50)
        print(f"📥 درخواست دریافت شد: {request}")
        
        # پردازش درخواست از طریق Middleware‌ها
        for middleware in self.middlewares:
            request = middleware.process_request(request)
            if request.get('cached_response'):
                return request['cached_response']
        
        # پردازش اصلی (Handler)
        response = self._handle(request)
        
        # پردازش پاسخ از طریق Middleware‌ها (به ترتیب معکوس)
        for middleware in reversed(self.middlewares):
            response = middleware.process_response(response)
        
        print(f"📤 پاسخ ارسال شد: {response}")
        print("="*50)
        return response
    
    def _handle(self, request: Dict) -> Dict:
        """Handler اصلی"""
        path = request.get('path', '/')
        return {
            'path': path,
            'status': 200,
            'data': f"Response for {path}",
            'user': request.get('user', 'anonymous')
        }


# استفاده
server = Server()

# افزودن Middleware‌ها
server.add_middleware(LoggingMiddleware())
server.add_middleware(AuthenticationMiddleware())
server.add_middleware(RateLimitMiddleware(5))
server.add_middleware(CacheMiddleware())

# تست درخواست‌ها
try:
    server.handle_request({'path': '/api/users', 'token': 'abc123'})
    server.handle_request({'path': '/api/users', 'token': 'abc123'})  # از کش
    server.handle_request({'path': '/api/posts', 'token': 'abc123'})
    
    # درخواست بدون توکن
    # server.handle_request({'path': '/api/users'})  # ❌ PermissionError
    
except Exception as e:
    print(f"خطا: {e}")
```

```python
from abc import ABC, abstractmethod
from typing import List, Dict, Callable, Any
from datetime import datetime

# رویداد انتزاعی
class Event(ABC):
    @abstractmethod
    def get_type(self) -> str:
        """نوع رویداد"""
        pass
    
    @abstractmethod
    def get_data(self) -> Dict:
        """داده‌های رویداد"""
        pass


# شنونده انتزاعی
class EventListener(ABC):
    @abstractmethod
    def on_event(self, event: Event) -> None:
        """پردازش رویداد"""
        pass


# رویدادهای مختلف
class UserCreatedEvent(Event):
    def __init__(self, user_id: int, username: str):
        self.user_id = user_id
        self.username = username
        self.timestamp = datetime.now()
    
    def get_type(self) -> str:
        return "USER_CREATED"
    
    def get_data(self) -> Dict:
        return {
            'user_id': self.user_id,
            'username': self.username,
            'timestamp': self.timestamp.isoformat()
        }


class OrderPlacedEvent(Event):
    def __init__(self, order_id: int, amount: float):
        self.order_id = order_id
        self.amount = amount
        self.timestamp = datetime.now()
    
    def get_type(self) -> str:
        return "ORDER_PLACED"
    
    def get_data(self) -> Dict:
        return {
            'order_id': self.order_id,
            'amount': self.amount,
            'timestamp': self.timestamp.isoformat()
        }


# شنوندگان مختلف
class EmailNotificationListener(EventListener):
    def on_event(self, event: Event) -> None:
        data = event.get_data()
        print(f"📧 ارسال ایمیل برای رویداد {event.get_type()}: {data}")


class LoggingListener(EventListener):
    def on_event(self, event: Event) -> None:
        print(f"📝 ثبت در لاگ: {event.get_type()} - {event.get_data()}")


class AnalyticsListener(EventListener):
    def on_event(self, event: Event) -> None:
        print(f"📊 ارسال به آنالیتیکس: {event.get_type()}")


class AuditListener(EventListener):
    def on_event(self, event: Event) -> None:
        print(f"🔍 ثبت در حسابرسی: {event.get_type()}")


# سیستم رویداد
class EventBus:
    def __init__(self):
        self.listeners: Dict[str, List[EventListener]] = {}
    
    def subscribe(self, event_type: str, listener: EventListener) -> None:
        """اشتراک در رویداد"""
        if event_type not in self.listeners:
            self.listeners[event_type] = []
        self.listeners[event_type].append(listener)
        print(f"✅ {listener.__class__.__name__} در {event_type} مشترک شد")
    
    def unsubscribe(self, event_type: str, listener: EventListener) -> None:
        """لغو اشتراک"""
        if event_type in self.listeners:
            self.listeners[event_type] = [
                l for l in self.listeners[event_type] 
                if l != listener
            ]
    
    def publish(self, event: Event) -> None:
        """انتشار رویداد"""
        event_type = event.get_type()
        print(f"\n📢 انتشار رویداد: {event_type}")
        
        if event_type in self.listeners:
            for listener in self.listeners[event_type]:
                listener.on_event(event)


# استفاده
event_bus = EventBus()

# اشتراک شنوندگان
event_bus.subscribe("USER_CREATED", EmailNotificationListener())
event_bus.subscribe("USER_CREATED", LoggingListener())
event_bus.subscribe("USER_CREATED", AnalyticsListener())
event_bus.subscribe("USER_CREATED", AuditListener())

event_bus.subscribe("ORDER_PLACED", EmailNotificationListener())
event_bus.subscribe("ORDER_PLACED", LoggingListener())
event_bus.subscribe("ORDER_PLACED", AnalyticsListener())

# انتشار رویدادها
user_event = UserCreatedEvent(1, "ali123")
event_bus.publish(user_event)

order_event = OrderPlacedEvent(100, 500000)
event_bus.publish(order_event)
```

```python
from abc import ABC, abstractmethod
from functools import wraps

def validate_input(func):
    @wraps(func)
    def wrapper(self, *args, **kwargs):
        for arg in args:
            if not isinstance(arg, (int, float, str)):
                raise ValueError("ورودی نامعتبر!")
        return func(self, *args, **kwargs)
    return wrapper

class Calculator(ABC):
    @abstractmethod
    @validate_input
    def calculate(self, a, b):
        pass

class Adder(Calculator):
    def calculate(self, a, b):
        return a + b

calc = Adder()
print(calc.calculate(5, 3))  # 8
# calc.calculate("a", 3)  # ❌ ValueError
```

```python
from abc import ABC, abstractmethod
from typing import TypeVar, Generic, List

T = TypeVar('T')

class Repository(ABC, Generic[T]):
    @abstractmethod
    def add(self, item: T) -> None:
        pass
    
    @abstractmethod
    def get_all(self) -> List[T]:
        pass
    
    @abstractmethod
    def find_by_id(self, id: int) -> T:
        pass

class User:
    def __init__(self, id: int, name: str):
        self.id = id
        self.name = name

class UserRepository(Repository[User]):
    def __init__(self):
        self.users: List[User] = []
    
    def add(self, item: User) -> None:
        self.users.append(item)
    
    def get_all(self) -> List[User]:
        return self.users
    
    def find_by_id(self, id: int) -> User:
        for user in self.users:
            if user.id == id:
                return user
        raise ValueError("کاربر یافت نشد")

repo = UserRepository()
repo.add(User(1, "علی"))
print(repo.get_all())
```
