```python
# define class
class Person:
    def __init__(self, name, city, age=65):
        self.name = name
        self.age = age
        self.city = city

# define object
name = input("you'r name:")
age = input("you'r age:")
city = input("you'r city:")
p = Person(name, city, age)

print("you'r name is {}, age is {}, city is {}".format(p.name, p.age, p.city))
```

```python
class Cars:
    def __init__(self, brand, country):
        self.brand = brand
        self.country = country

car = Cars("BMW", "germany")
print(car.brand)
print(car.country)
```

```python
#تابع ()__str__
class Cars02:
    def __init__(self, brand, country):
        self.brand= brand
        self.country= country
    
    def __str__(self):
        return f"{self.brand}({self.country})"

car1 = Cars02("BMW", "germany")
print(car1)
```

```python
# change value from object of vvariable
class Cars:
  def __init__(self, brand, score):
    self.brand = brand
    self.score = score

car1 = Cars("BMW", 60)

car1.score = 10
print(car1.score)
```

```python
# delete value from object of variable
class Cars:
  def __init__(self, brand, score):
    self.brand = brand
    self.score = score

car1 = Cars("BMW", 60)
del car1.score
print(car1.score)
```
