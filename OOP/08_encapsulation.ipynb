```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner              # عمومی
        self.__balance = balance        # خصوصی (دو آندرلاین)
    
    # Getter: متد برای خواندن مقدار خصوصی
    def get_balance(self):
        return self.__balance
    
    # Setter: متد برای تغییر مقدار خصوصی با کنترل
    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print(f"موجودی به {amount} تومان تغییر کرد.")
        else:
            print("❌ موجودی نمی‌تواند منفی باشد!")
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"✅ {amount} تومان واریز شد. موجودی: {self.__balance}")
        else:
            print("❌ مبلغ واریزی باید مثبت باشد!")
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"✅ {amount} تومان برداشت شد. موجودی: {self.__balance}")
        else:
            print("❌ مبلغ نامعتبر یا موجودی کافی نیست!")


# استفاده
account = BankAccount("علی", 1000)

# ✅ روش صحیح: استفاده از متدها
print(f"موجودی فعلی: {account.get_balance()}")  # 1000
account.deposit(500)      # ✅ 1500
account.withdraw(200)     # ✅ 1300

# ❌ تلاش برای دسترسی مستقیم به متغیر خصوصی
# print(account.__balance)  # AttributeError!

# ❌ تلاش برای تغییر مستقیم (این کار نمی‌کند!)
account.__balance = 999999
print(f"موجودی بعد از تلاش: {account.get_balance()}")  # هنوز 1300!

# ✅ روش صحیح برای تغییر
account.set_balance(2000)  # ✅ موجودی به 2000 تغییر کرد.
```

```python
import hashlib

class User:
    def __init__(self, username, password):
        self.username = username
        self.__password = self.__hash_password(password)  # رمز را هش می‌کنیم
    
    def __hash_password(self, password):
        """متد خصوصی برای هش کردن رمز عبور"""
        return hashlib.sha256(password.encode()).hexdigest()
    
    def check_password(self, password):
        """بررسی صحت رمز عبور"""
        return self.__password == self.__hash_password(password)
    
    def change_password(self, old_password, new_password):
        """تغییر رمز عبور با بررسی رمز قدیمی"""
        if self.check_password(old_password):
            self.__password = self.__hash_password(new_password)
            print("✅ رمز عبور با موفقیت تغییر کرد.")
        else:
            print("❌ رمز عبور قدیمی اشتباه است!")
    
    def get_username(self):
        return self.username


# استفاده
user = User("ali123", "MySecretPass123")

# ✅ بررسی رمز
print(user.check_password("MySecretPass123"))  # True
print(user.check_password("WrongPass"))         # False

# ✅ تغییر رمز
user.change_password("MySecretPass123", "NewPass456")

# ❌ نمی‌توانیم رمز هش شده را ببینیم
# print(user.__password)  # AttributeError!
```

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.__price = price
        self.__stock = stock
    
    @property
    def price(self):
        """Getter با استفاده از @property"""
        return self.__price
    
    @price.setter
    def price(self, value):
        """Setter با استفاده از @property"""
        if value > 0:
            self.__price = value
        else:
            print("❌ قیمت باید مثبت باشد!")
    
    @property
    def stock(self):
        return self.__stock
    
    def sell(self, quantity):
        if 0 < quantity <= self.__stock:
            self.__stock -= quantity
            print(f"✅ {quantity} عدد {self.name} فروخته شد. موجودی: {self.__stock}")
        else:
            print("❌ تعداد نامعتبر یا موجودی کافی نیست!")


# استفاده
product = Product("لپ‌تاپ", 50000000, 10)

# ✅ استفاده از property مثل متغیر
print(f"قیمت: {product.price}")  # 50000000
product.price = 45000000         # ✅ تغییر قیمت
# product.price = -100           # ❌ قیمت باید مثبت باشد!

# ✅ فروش محصول
product.sell(3)  # ✅ 3 عدد لپ‌تاپ فروخته شد. موجودی: 7

# ❌ نمی‌توانیم مستقیم تغییر دهیم
# product.__stock = 100  # AttributeError!
```
