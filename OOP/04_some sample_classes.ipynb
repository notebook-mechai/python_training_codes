```python
class Car:
    def __init__(self, brand, color, speed):
        self.brand = brand
        self.color = color
        self.speed = speed

    def honk(self):
        print(f"{self.brand}")

    def accelerate(self):
        self.speed += 20
        print(f"{self.brand} الان با سرعت {self.speed} حرکت می‌کند.")

car1 = Car("پراید", "سفید", 0)
car2 = Car("بنز", "مشکی", 0)
car3 = Car("پژو", "قرمز", 0)

car1.honk()          # پراید بوق می‌زند: بوق بوق!
car2.accelerate()    # بنز الان با سرعت 20 حرکت می‌کند.
car3.accelerate()    # پژو الان با سرعت 20 حرکت می‌کند.
car3.accelerate()    # پژو الان با سرعت 40 حرکت می‌کند. (مستقل از بقیه)
```

```python
class Student:

    def __init__(self, name, student_id, grade):
        self.name = name
        self.student_id = student_id
        self.grade = grade

    def introduce(self):
        print(f"سلام! من {self.name} هستم با شماره دانشجویی {self.student_id}")

    def is_passed(self):
        if self.grade >= 10:
            print(f"{self.name} قبول شده است.")
        else:
            print(f"{self.name} مردود شده است.")

student1 = Student("علی", 1001, 18)
student2 = Student("مریم", 1002, 8)
student3 = Student("رضا", 1003, 15)

# استفاده از متدها
student1.introduce()
student1.is_passed()

student2.introduce()
student2.is_passed()
```

```python
class BusinessCard:
    def __init__(self, person_name, job, phone):
        self.person_name = person_name
        self.job = job
        self.phone = phone

    def show_info(self):
        print(f"نام: {self.person_name} | شغل: {self.job} | تلفن: {self.phone}")

card1 = BusinessCard("دکتر احمدی", "پزشک", "09121111111")
card2 = BusinessCard("مهندس کریمی", "برنامه‌نویس", "09122222222")

card1.show_info()
card2.show_info()
```
