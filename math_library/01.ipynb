```python
# factorial with while
# 4! = 1*2*3*4
n = 4
i = 1
result = 1
if n == 0 or n == 1:
    print("The factorial {} result is: {}".format(n,result))
else:
    while(i <= n):
        result *= i
        i = i + 1
    print("The factorial {} result is: {}".format(n,result))
```

```python
# factorial with for
# 4! = 1*2*3*4
n = target = result = 0 # reset all value variables

n = 4
target = n +1
result = 1
if n == 0 or n == 1:
    print("The factorial {} result is: {}".format(n,result))
else:
    for i in range(1, target):
        result *= i
    print("The factorial {} result is: {}".format(n,result))
```
