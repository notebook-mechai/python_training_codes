```python
a = 2
b = 8
print("a is: {} , b is: {}".format(a,b))

temp = a
a = b
b = temp
print("after swap numbers, a is: {} , b is: {}".format(a,b))
```

```python
def change_nums(a, b):
    temp = a
    a = b
    b = temp
    print("after swap numbers, a is: {} , b is: {}".format(a,b))

a = 12
b = 18
print("a is: {} , b is: {}".format(a,b))
change_nums(a, b)   
```

```python
def ch_nums(x,y):
    temp = x
    x = y
    y = temp
    return x,y

(m,n) = ch_nums(65, 10)
print("m is: {} , n is: {}".format(m,n))
```
