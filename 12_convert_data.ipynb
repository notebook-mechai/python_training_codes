```python
print("\n------- concept compare numbers")

value = 4
value_01 = "4"
value_02 = '4'

print("type of value variable:", type(value))
print("type of value_01 variable:", type(value_01))
print("type of value_02 variable:", type(value_02))

if (value == value_01):
    print("\n4 number is equal 4 string")
else:
    print("\n4 number IS NOT EQUAL 4 string")


if (value_01 == value_02):
    print("\n4 string is equal 4 char")
else:
    print("\n4 string IS NOT EQUAL 4 char")


if (value == int(value_01)):
    print("\n4 number is equal 4 string after convert")
else:
    print("\n4 number IS NOT EQUAL 4 string after convert")
```
